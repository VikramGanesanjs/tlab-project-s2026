```python
full = nib.load('/common/ganesanv/tlab/data/tcia/duke_breast_cancer_processed/Breast_MRI_007/pre/volume.nii')
left = nib.load('/common/ganesanv/tlab/data/tcia/duke_breast_cancer_processed/Breast_MRI_007/pre/left.nii')
right = nib.load('/common/ganesanv/tlab/data/tcia/duke_breast_cancer_processed/Breast_MRI_007/pre/right.nii')

full_data = full.get_fdata()
left_data = left.get_fdata()
right_data = right.get_fdata()

plt.imshow(full_data[:, :, 80], cmap=plt.cm.bone)
plt.show()

plt.imshow(left_data[:, :, 80], cmap=plt.cm.bone)
plt.show()

plt.imshow(right_data[:, :, 80], cmap=plt.cm.bone)
plt.show()
```

```python
masks = nib.load('/common/ganesanv/tlab/data/tcia/duke_breast_cancer_mri/Breast_MRI_002/1.3.6.1.4.1.14519.5.2.1.29344851079648912610491979642001151972/MR_1.3.6.1.4.1.14519.5.2.1.92459963600394245252521533321918466447/breast_divider.nii.gz')
print(masks.shape[2] == full_data.shape[2], full_data.shape, masks.shape)
m = masks.get_fdata()
for i in range(3):
    m = np.rot90(m)
m = np.flip(m, axis=1)
m = np.flip(m, axis=2)
fig, ax = plt.subplots(1, 1)

ax.imshow(m[:, :, 71], cmap=plt.cm.bone)
ax.imshow(full_data[:, :, 71], cmap=plt.cm.bone, alpha=0.5)
plt.show()
```
